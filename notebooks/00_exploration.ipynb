Quick EDA for the Sackmann point-by-point data. Use this to sanity-check
coverage and base rates before running the experiments. Nothing here is a

Run `python -m src.data_loading` first to create `data/processed/points.parquet`.

```python
import sys; sys.path.insert(0, '..')
import pandas as pd
from src.config import PROCESSED_DIR

pts = pd.read_parquet(PROCESSED_DIR / 'points.parquet')
print(len(pts), 'points;', pts.match_id.nunique(), 'matches')
pts.head()
```

```python
# Server win rate (the baseline momentum must beat)
print('P(server wins point) =', pts['server_won'].mean().round(4))
# Naive 'won last point -> win next' association (BEWARE: confounded by serve)
pts['prev_p1'] = pts.groupby('match_id')['p1_won'].shift(1)
print(pts.dropna(subset=['prev_p1']).groupby('prev_p1')['p1_won'].mean())
```
